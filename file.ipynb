```python
import torch
from langchain.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import Chroma
from langchain.llms import LlamaCpp
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationalRetrievalChain
from langchain.prompts import PromptTemplate
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

```python
llm_answer_gen = LlamaCpp(
    streaming=True,
    model_path=r"./mistral-7b-openorca.Q4_0.gguf",
    temperature=0.75,
    top_p=1,
    f16_kv=True,
    verbose=False,
    n_ctx=4096
)
```

```python
# Update the model_path to the downloaded file
llm_answer_gen = LlamaCpp(
    streaming=True,
    model_path="./mistral-7b-openorca.Q4_0.gguf",
    temperature=0.75,
    top_p=1,
    f16_kv=True,
    verbose=False,
    n_ctx=4096
)
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2", model_kwargs={"device": device})
```

```python
# vector which contains the chunks
loader = PyPDFLoader("FLNonIID.pdf")
data = loader.load()
```

```python
data
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
text_splitter
```

```python
text_chunks = text_splitter.split_documents(data)
text_chunks
```

```python
vector_store = Chroma.from_documents(text_chunks, embeddings)
```

```python
memory = ConversationBufferMemory(memory_key="chat_history", return_messages=True, output_key="answer")
```

```python
#Again to my part:
custom_prompt=PromptTemplate(
    input_variables=["context", "question"],
    template="""Answer using only the provided context. If the context does not contain the answer, say so."
            Context:
            {context}

            Question:
            {question}

            Answer:
"""
)
```

```python
answer_gen_chain = ConversationalRetrievalChain.from_llm(
    llm=llm_answer_gen,
    retriever=vector_store.as_retriever(search_kwargs={"k": 5}),
    memory=memory,
    combine_docs_chain_kwargs={"prompt": custom_prompt},
    return_source_documents=True,
    output_key="answer"
)
```

```python
while True:
    user_input = input("Enter a question (or 'q' to quit): ")
    if user_input.lower() == 'q':
        break

    result = answer_gen_chain({"question": user_input})
    print("\nAnswer:", result["answer"])
    print("\nSources:")
    for doc in result["source_documents"]:
        page = doc.metadata.get("page", "?")
        snippet = doc.page_content[:200]
        print(f"  [page {page}] {snippet}")
```
